# PARC2026 Colab A100 preflight
Prepare a disposable A100 workspace before running model comparisons. Large caches and datasets stay under `/content`; only experiment definitions/results belong in git.

```python
import os, platform, subprocess, sys
print('python:', sys.version)
print('platform:', platform.platform())
subprocess.run(['nvidia-smi'], check=True)
```

```python
from pathlib import Path
ROOT = Path('/content/parc2026')
for p in [ROOT, ROOT/'vendor', ROOT/'cache', ROOT/'datasets', ROOT/'outputs']:
    p.mkdir(parents=True, exist_ok=True)
print(ROOT)
```

```python
from pathlib import Path
repo = Path('/content/parc2026/py_AI')
if not repo.exists():
    subprocess.run(['git','clone','https://github.com/yu37330/py_AI.git', str(repo)], check=True)
else:
    subprocess.run(['git','-C',str(repo),'fetch','--all','--prune'], check=True)
print(subprocess.check_output(['git','-C',str(repo),'rev-parse','HEAD'], text=True).strip())
```
